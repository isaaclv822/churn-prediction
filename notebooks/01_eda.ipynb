```python
(df['Churn'] == 'Yes').groupby(df['Contract']).mean()
```

- Parmi les clients en contrat mensuel, **42,7%** sont partis. 
- Parmi ceux en contrat d’un an, **11,3%** sont partis. 
- Parmi ceux en contrat de deux ans, **2,8%** sont partis.   
Conclusion : un client en contrat mensuel part environ **15 fois plus souvent** (42,7 / 2,8) qu'un client engagé sur deux ans.

**Constat 1 : le contrat**. 42,7 % des clients en contrat mensuel sont partis, contre 11,3 % en contrat d’un an et 2,8 % en contrat de deux ans.

```python
import seaborn as sns

sns.boxplot(data=df, x='Churn', y='tenure')
```

Pour les clients partis (Yes), beaucoup ont une ancienneté très courte, avec quelques clients partis tard qui étirent la boîte vers le haut. Pour les clients restés (No), les ancienneté sont étalées de façon plus régulière sur toute la plage.  
A retenir : **les départs se concentrent dans les premiers mois.**

```python
df.groupby('Churn')['tenure'].describe()
```

**Constat 2 : l’ancienneté**. La médiane d’ancienneté est de 10 mois pour les clients partis contre 38 mois pour les clients restés. La moitié des départs a lieu avant 10 mois et un quart avant 2 mois.
